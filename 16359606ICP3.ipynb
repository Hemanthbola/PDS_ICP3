```python
import pandas as pd

# Loading the data
data_url ='https://raw.githubusercontent.com/JuliaData/CSV.jl/refs/heads/main/test/testfiles/Sacramentorealestatetransactions.csv'
df = pd.read_csv(data_url)


```

Find the number of the columns in the data frame

```python
num_columns = df.shape[1]
print(f"Number of columns in the DataFrame: {num_columns}")
```

Find the mean of the house prices

```python
mean_price = df['price'].mean()
print(f"Mean house price: {mean_price:.2f}")
```

Find the median of the house prices

```python
median_price = df['price'].median()
print(f"Median house price: {median_price:.2f}")
```

Find the different types of property in the dataset

```python
property_types = df['type'].unique()
print("Different property types:", property_types)
```

Find the maximum price of a "condo" in the dataset


```python
max_condo_price = (
    df
    .loc[df['type'].str.strip().str.lower() == 'condo', 'price']
    .max()
)
print(f"\nMaximum price of a condo: {max_condo_price:.2f}")
```

Find the mean price of each type of property in the dataset

```python
df['type'] = df['type'].str.strip().str.lower()
mean_prices_by_type = df.groupby('type')['price'].mean().round(2)

# Displaying results
print("Mean price by property type:")
print(mean_prices_by_type)
```

Find the count of different property types in dataset

```python
type_counts = df['type'].value_counts()

# Displaying the counts
print("Counts by property type:")
print(type_counts)
```
